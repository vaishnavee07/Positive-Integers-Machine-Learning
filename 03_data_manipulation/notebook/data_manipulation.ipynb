```python
##merging and joining dataframes

##creating df

df1 = pd.DataFrame({'key':['A','B','C'], 'value1' : [1,2,3]})
df2 = pd.DataFrame({'key':['A','B','D'], 'value1' : [4,5,6]})
print(df1)
print(df2)
```

```python
##merging df on key columns

##INNER JOIN

pd.merge(df1,df2,on="key",how="inner")
```

```python
##OUTER join

pd.merge(df1,df2,on="key",how="outer")
```

```python
##LEFT OUTER JOIN

pd.merge(df1,df2,on="key",how="left")
```

```python
##RIGHT OUTER JOIN

pd.merge(df1,df2,on="key",how="right")
```

```python
##date time conversion

sales_df["ORDERDATE"] = pd.to_datetime(
    sales_df["ORDERDATE"]
)

sales_df["ORDERDATE"].head()
```

```python
sales_df["ORDERDATE"].dtype
```

```python
sales_df["Day"] = sales_df["ORDERDATE"].dt.day
```

```python

sales_df["Weekday"] = sales_df["ORDERDATE"].dt.day_name()
```

```python

sales_df["Month_Name"] = sales_df["ORDERDATE"].dt.month_name()
```

```python
sales_df[
    ["ORDERDATE","Day","Weekday","Month_Name"]
].head()
```

```python
##feature engineering

sales_df["Revenue_Per_Item"] = (
    sales_df["SALES"] /
    sales_df["QUANTITYORDERED"]
)

sales_df[
    ["SALES",
     "QUANTITYORDERED",
     "Revenue_Per_Item"]
].head()
```

```python
##outlier detection

Q1 = sales_df["SALES"].quantile(0.25)

Q3 = sales_df["SALES"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = sales_df[
    (sales_df["SALES"] < lower)
    |
    (sales_df["SALES"] > upper)
]

outliers.head()
```
