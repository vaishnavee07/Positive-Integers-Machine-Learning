Sales data visualization

```python
sales_df = pd.read_csv(
    "../data/sales_data.csv",
    encoding="latin1"
)
sales_df.head(5)
```

```python
##sales by country

sales_country = sales_df.groupby("COUNTRY")["SALES"].sum().sort_values(
    ascending=False
).head(10)
plt.figure(figsize=(10,5))
sales_country.plot(kind="bar")
plt.title("Top 10 Countries by Sales")
plt.ylabel("Sales")
plt.show()
```

```python
##sales by product line

sales_product = sales_df.groupby("PRODUCTLINE")["SALES"].sum()
plt.figure(figsize=(10,5))
sales_product.plot(kind="bar")
plt.title("Sales by Product Line")
plt.show()
```

```python
##monthly sales

monthly_sales = sales_df.groupby("MONTH_ID")["SALES"].sum()
plt.figure(figsize=(10,5))
monthly_sales.plot()
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
##deal size distribution

deal_size = sales_df["DEALSIZE"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(deal_size,labels=deal_size.index,autopct="%1.1f%%")
plt.title("Deal Size Distribution")
plt.show()
```
